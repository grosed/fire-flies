#### Notes

Had to add fuse to run installer

```bash
sudo add-apt-repository universe
sudo apt install libfuse2
```

May have to modify udev rules. See [here](https://support.arduino.cc/hc/en-us/articles/9005041052444-Fix-udev-rules-on-Linux) and [here](https://docs.arduino.cc/software/ide-v2/tutorials/getting-started/ide-v2-downloading-and-installing/)

![title](./notebooks/images/usb-port-issue.png)

````
Sketch uses 444 bytes (1%) of program storage space. Maximum is 30720 bytes.
Global variables use 9 bytes (0%) of dynamic memory, leaving 2039 bytes for local variables. Maximum is 2048 bytes.
avrdude: ser_open(): can't open device "/dev/ttyUSB0": Permission denied
Failed uploading: uploading error: exit status 1
````

Solved the above by allowing writing to USB

```bash
sudo chmod a+rw /dev/ttyUSB0
```
